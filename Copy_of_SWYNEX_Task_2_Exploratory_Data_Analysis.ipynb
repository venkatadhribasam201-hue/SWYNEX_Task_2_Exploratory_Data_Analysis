```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/cleaned_marketing_campaign.csv')
```

```python
print("Average Income:", round(df['Income'].mean(), 2))
print("Average Recency:", round(df['Recency'].mean(), 2))
print("Average Web Purchases:", round(df['NumWebPurchases'].mean(), 2))
print("Average Store Purchases:", round(df['NumStorePurchases'].mean(), 2))
print("Average Catalog Purchases:", round(df['NumCatalogPurchases'].mean(), 2))
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Income'], bins=30)
plt.title('Income Distribution')
plt.xlabel('Income')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(8,5))
df['Education'].value_counts().plot(kind='bar')
plt.title('Customers by Education Level')
plt.xlabel('Education')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,5))
df['Response'].value_counts().plot(kind='bar')
plt.title('Campaign Response')
plt.xlabel('Response (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
spending = df[
    ['MntWines',
     'MntFruits',
     'MntMeatProducts',
     'MntFishProducts',
     'MntSweetProducts',
     'MntGoldProds']
].sum()

plt.figure(figsize=(10,5))
spending.plot(kind='bar')
plt.title('Total Spending by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Spending')
plt.xticks(rotation=45)
plt.show()
```

```python
channels = df[
    ['NumWebPurchases',
     'NumCatalogPurchases',
     'NumStorePurchases']
].sum()

plt.figure(figsize=(8,5))
channels.plot(kind='bar')
plt.title('Total Purchases by Channel')
plt.xlabel('Purchase Channel')
plt.ylabel('Total Purchases')
plt.xticks(rotation=0)
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(14,10))
sns.heatmap(numeric_df.corr(), cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
